```python
import numpy 
import cv2
import libs_common
import rl_libs 
```

# create environment wrapper

- 128 parallel envs
- env.step(actions), returns : states, rewards, dones

- actions.shape = (n_envs, ),   discrete actions, 4
- state.shape   = (n_envs, 8),   8 dims state space for each environment
- rewards.shape = (n_envs, ),   scalar reward value, float
- dones.shape = (n_envs, ),     scaler bool value if episode done

```python
n_envs = 128
envs = libs_common.VecEnv(n_envs=n_envs) 
```

```python
# quick smoke test
obs = envs.reset()

print("obs batch:", obs.shape)          # (16, 115)
print("actions:", envs.action_space)    # Discrete(19)

for t in range(2000):
    actions = numpy.random.randint(0, envs.action_space.n, size=n_envs)
    obs, reward, done = envs.step(actions)
    if not envs.render(env_id=0, wait_ms=20):              # Esc to stop
        break

cv2.destroyAllWindows()
```
