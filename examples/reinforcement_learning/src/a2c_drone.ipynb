```python
import time
import numpy as np
import pybullet as p  # <--- Import pybullet
from gym_pybullet_drones.envs.HoverAviary import HoverAviary
from gym_pybullet_drones.envs.MultiHoverAviary import MultiHoverAviary
from gym_pybullet_drones.utils.enums import ObservationType, ActionType

# 1. Initialize environment with GUI rendering enabled
env = MultiHoverAviary(
    gui=True,
    obs=ObservationType.KIN,
    act=ActionType.RPM,
    num_drones=4
)

# -------------------------------------------------------------
# Hide standard PyBullet GUI sidebars, sliders, and extra views
# -------------------------------------------------------------
p.configureDebugVisualizer(p.COV_ENABLE_GUI, 0, physicsClientId=env.CLIENT)
p.configureDebugVisualizer(p.COV_ENABLE_RGB_BUFFER_PREVIEW, 0, physicsClientId=env.CLIENT)
p.configureDebugVisualizer(p.COV_ENABLE_DEPTH_BUFFER_PREVIEW, 0, physicsClientId=env.CLIENT)
p.configureDebugVisualizer(p.COV_ENABLE_SEGMENTATION_MARK_PREVIEW, 0, physicsClientId=env.CLIENT)

# 2. Reset the environment
obs, info = env.reset()

print("Action Space:", env.action_space)
print("Observation Space Shape:", obs.shape)

step_delay = 1.0 / env.CTRL_FREQ

try:
    for step in range(1000):
        actions = env.action_space.sample()
        obs, reward, terminated, truncated, info = env.step(actions)

        time.sleep(step_delay)

        print(reward)

        if terminated or truncated:
            print(f"Episode finished at step {step}")
            obs, info = env.reset()

finally:
    env.close()
```
