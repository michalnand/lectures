```python
import libs_common
import rl_libs 

import numpy 

n_envs = 1
envs = libs_common.VecEnv(n_envs=n_envs)

rl_agent = rl_libs.A2CAgent(
    n_envs=n_envs,
    state_shape=(8,),
    n_actions=4,
    Model=rl_libs.MLPRLModel,
    learning_rate=0.002,
    n_steps=128,    
    gamma=0.99,
    entropy_beta=0.01,
    device="cpu",
)


rl_agent.load("results/a2c_lunar_lander/")


# init variables for agent
states  = envs.reset()
rewards = numpy.zeros(n_envs, dtype=numpy.float32)
dones   = numpy.zeros(n_envs, dtype=bool)


for n in range(1000):

    actions = rl_agent.step(states, rewards, dones, True)

    states, rewards, dones = envs.step(actions)

    envs.render()

    if dones[0]:
        states[0] = envs.reset(0)
        print("done")

```
